```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age 
        
    def show_details(self):
        print("Person Details")
        print(f"Name      : {self.name}")
        print(f"Age       : {self.age}")

class Employee(Person):
    def __init__(self, name, age, id, salary):
        super().__init__(name, age)
        self.id = id
        self.salary=salary

    def employee_details(self):
        print(f"ID : {self.id}")
        print(f"Salary : {self.salary}")

class Manager(Person):
    def __init__(self, name, age, id, salary,department):
        super().__init__(name, age)
        self.department = department

    def manager_details(self):
        print(f"Department  : {self.department}")
class Developer(Person):
    def __init__(self, name, age, id, salary, emp_id, department):
        super().__init__(name, age)
        self.department = department

print("1. create a person")
print("2. create an employee")
print("3. create a manager")
print("4. show details")
print("5. exit")

n=int(input("Enter your choice = "))

match n:
    case 1:
        a=input("enter your name =")
        b=int(input("enter your age ="))
        Person(a,b)
    case 2:
        a=input("enter your name =")
        b=int(input("enter your age ="))
        c=int(input("enter your id="))
        d=int(input("enter your salary"))
        Employee(a,b,c,d)
    case 3:
        a=input("enter your name =")
        b=int(input("enter your age ="))
        c=int(input("enter your id="))
        d=int(input("enter your salary"))
        e=input("enter your  department")
        Manager(a,b,c,d,e)
    case 4:
        a=input("enter your name =")
        b=int(input("enter your age ="))
        c=int(input("enter your id="))
        d=int(input("enter your salary"))
        show_details(a,b,c,d)
    case 5:
        print("Exit")
        print("Exiting the system. all resource have been freed")
        print("Goodbye!")
        exit()
```
